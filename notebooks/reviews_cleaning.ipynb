```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd()
)

RAW_DATA_DIR = PROJECT_ROOT / "Data" / "raw"
CLEANED_DATA_DIR = PROJECT_ROOT / "Data" / "cleaned"

CLEANED_DATA_DIR.mkdir(parents=True, exist_ok=True)

REVIEWS_FILE = RAW_DATA_DIR / "olist_order_reviews_dataset.csv"

df = pd.read_csv(REVIEWS_FILE)
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

display(df.head())
```

```python
# Check missing values

missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

display(missing_report)
```

```python
# Check exact duplicates

print("Exact Duplicate Rows:", df.duplicated().sum())

print("Duplicate Review IDs:", df["review_id"].duplicated().sum())

print("Duplicate Order IDs:", df["order_id"].duplicated().sum())

print(
    "Duplicate Review-Order Pairs:",
    df.duplicated(subset=["review_id", "order_id"]).sum()
)
```

```python
# Review score distribution

print("Review Score Distribution:")

display(
    df["review_score"]
    .value_counts(dropna=False)
    .sort_index()
    .to_frame("Count")
)

print("\nReview Score Statistics:")
display(df["review_score"].describe())
```

```python
# Check invalid review scores

invalid_scores = df[
    ~df["review_score"].isin([1, 2, 3, 4, 5])
]

print("Invalid Review Scores:", len(invalid_scores))

display(invalid_scores)
```

```python
# Inspect duplicated review IDs

duplicate_review_ids = df[
    df["review_id"].duplicated(keep=False)
].copy()

duplicate_review_ids = duplicate_review_ids.sort_values(
    ["review_id", "order_id"]
)

print("Rows with Duplicate Review IDs:", len(duplicate_review_ids))
print("Unique Duplicated Review IDs:",
      duplicate_review_ids["review_id"].nunique())

display(duplicate_review_ids.head(30))
```

```python
# Inspect orders with multiple reviews

duplicate_orders = df[
    df["order_id"].duplicated(keep=False)
].copy()

duplicate_orders = duplicate_orders.sort_values(
    ["order_id", "review_creation_date"]
)

print("Rows with Duplicate Orders:", len(duplicate_orders))
print("Unique Orders with Multiple Reviews:",
      duplicate_orders["order_id"].nunique())

display(duplicate_orders.head(30))
```

```python
# Analyze duplicated review IDs

duplicate_review_analysis = (
    duplicate_review_ids
    .groupby("review_id")
    .agg(
        occurrences=("review_id", "size"),
        unique_orders=("order_id", "nunique"),
        unique_scores=("review_score", "nunique"),
        unique_titles=("review_comment_title", "nunique"),
        unique_messages=("review_comment_message", "nunique"),
        unique_creation_dates=("review_creation_date", "nunique"),
        unique_answer_dates=("review_answer_timestamp", "nunique")
    )
    .reset_index()
)

print("Duplicate Review ID Groups:",
      len(duplicate_review_analysis))

print("\nOccurrences Distribution:")
display(
    duplicate_review_analysis["occurrences"]
    .value_counts()
    .sort_index()
)

print("\nReview IDs with Different Scores:")
display(
    duplicate_review_analysis[
        duplicate_review_analysis["unique_scores"] > 1
    ].head(10)
)

print("\nReview IDs with Different Comments:")
display(
    duplicate_review_analysis[
        (duplicate_review_analysis["unique_titles"] > 1) |
        (duplicate_review_analysis["unique_messages"] > 1)
    ].head(10)
)
```

```python
# Analyze orders with multiple reviews

order_review_analysis = (
    duplicate_orders
    .groupby("order_id")
    .agg(
        review_count=("review_id", "count"),
        unique_review_ids=("review_id", "nunique"),
        unique_scores=("review_score", "nunique"),
        unique_creation_dates=("review_creation_date", "nunique"),
        unique_answer_dates=("review_answer_timestamp", "nunique")
    )
    .reset_index()
)

print("Orders with Multiple Reviews:",
      len(order_review_analysis))

print("\nNumber of Reviews per Order:")
display(
    order_review_analysis["review_count"]
    .value_counts()
    .sort_index()
)

print("\nOrders with Different Scores:")
display(
    order_review_analysis[
        order_review_analysis["unique_scores"] > 1
    ].head(10)
)

print("\nOrders with Different Review Dates:")
display(
    order_review_analysis[
        order_review_analysis["unique_creation_dates"] > 1
    ].head(10)
)
```

```python
# Inspect review date columns

date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

date_check = df[date_columns].copy()

for col in date_columns:
    date_check[col] = pd.to_datetime(
        date_check[col],
        errors="coerce"
    )

print("Invalid Creation Dates:",
      date_check["review_creation_date"].isna().sum())

print("Invalid Answer Timestamps:",
      date_check["review_answer_timestamp"].isna().sum())

# Check answer timestamps before creation dates

date_issues = date_check[
    date_check["review_answer_timestamp"]
    < date_check["review_creation_date"]
]

print("\nAnswer Before Creation:", len(date_issues))

# Calculate response time

response_time = (
    date_check["review_answer_timestamp"]
    - date_check["review_creation_date"]
)

print("\nResponse Time Statistics:")
display(response_time.describe())

print("\nNegative Response Times:")
display(response_time[response_time < pd.Timedelta(0)].head(10))
```

```python
# Create a cleaned copy

cleaned_df = df.copy()

# Clean text columns
text_columns = [
    "review_id",
    "order_id",
    "review_comment_title",
    "review_comment_message"
]

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].astype("string").str.strip()

# Replace empty strings with missing values
comment_columns = [
    "review_comment_title",
    "review_comment_message"
]

for col in comment_columns:
    cleaned_df[col] = cleaned_df[col].replace("", pd.NA)

# Convert date columns
date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

for col in date_columns:
    cleaned_df[col] = pd.to_datetime(
        cleaned_df[col],
        errors="coerce"
    )

# Validate review scores
invalid_scores = cleaned_df[
    ~cleaned_df["review_score"].between(1, 5)
]

print("Original Shape:", df.shape)
print("Cleaned Shape:", cleaned_df.shape)
print("Invalid Scores:", len(invalid_scores))
print("Exact Duplicates:", cleaned_df.duplicated().sum())
print("Missing Values:")
display(cleaned_df.isna().sum())
```

```python
print("Original Rows:", len(df))
print("Cleaned Rows:", len(cleaned_df))

print("Invalid Scores:",
      (~cleaned_df["review_score"].between(1, 5)).sum())

print("Exact Duplicates:",
      cleaned_df.duplicated().sum())

print("Missing Review IDs:",
      cleaned_df["review_id"].isna().sum())

print("Missing Order IDs:",
      cleaned_df["order_id"].isna().sum())

print("Date Columns:")
print(cleaned_df[[
    "review_creation_date",
    "review_answer_timestamp"
]].dtypes)
```

```python
# 1. General quality report

quality_report = pd.DataFrame({
    "column": cleaned_df.columns,
    "dtype": cleaned_df.dtypes.astype(str).values,
    "missing_values": cleaned_df.isna().sum().values,
    "missing_percentage": (
        cleaned_df.isna().mean().values * 100
    ).round(2),
    "unique_values": cleaned_df.nunique().values
})

# 2. Duplicate review IDs report

duplicate_review_ids_report = cleaned_df[
    cleaned_df["review_id"].duplicated(keep=False)
].sort_values("review_id")

# 3. Multiple reviews per order report

multiple_reviews_report = cleaned_df[
    cleaned_df["order_id"].duplicated(keep=False)
].sort_values(["order_id", "review_creation_date"])

# 4. Long response time report

response_times = (
    cleaned_df["review_answer_timestamp"]
    - cleaned_df["review_creation_date"]
)

long_response_report = cleaned_df[
    response_times > pd.Timedelta(days=30)
].copy()

long_response_report["response_time_days"] = (
    response_times[response_times > pd.Timedelta(days=30)]
    .dt.total_seconds() / 86400
)

print("Quality Report:")
display(quality_report)

print("\nDuplicate Review ID Records:",
      len(duplicate_review_ids_report))

print("Orders with Multiple Reviews:",
      multiple_reviews_report["order_id"].nunique())

print("Responses Taking More Than 30 Days:",
      len(long_response_report))
```

```python
# Save cleaned dataset and quality reports

output_files = {
    "cleaned_reviews": CLEANED_DATA_DIR / "olist_order_reviews_cleaned.csv",
    "quality_report": CLEANED_DATA_DIR / "reviews_quality_report.csv",
    "duplicate_review_ids": CLEANED_DATA_DIR / "reviews_duplicate_review_ids.csv",
    "multiple_reviews": CLEANED_DATA_DIR / "reviews_multiple_reviews_per_order.csv",
    "long_response_times": CLEANED_DATA_DIR / "reviews_long_response_times.csv"
}

cleaned_df.to_csv(
    output_files["cleaned_reviews"],
    index=False
)

quality_report.to_csv(
    output_files["quality_report"],
    index=False
)

duplicate_review_ids_report.to_csv(
    output_files["duplicate_review_ids"],
    index=False
)

multiple_reviews_report.to_csv(
    output_files["multiple_reviews"],
    index=False
)

long_response_report.to_csv(
    output_files["long_response_times"],
    index=False
)

# Verify saved files

for name, path in output_files.items():
    print(f"{name}: {'Saved' if path.exists() else 'Missing'}")

# Verify cleaned dataset
saved_df = pd.read_csv(output_files["cleaned_reviews"])

print("\nSaved Dataset Shape:", saved_df.shape)
print("Expected Dataset Shape:", cleaned_df.shape)

assert saved_df.shape == cleaned_df.shape

print("Saved dataset verification passed.")
```
