```python
# Imports & Paths
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

RAW_DATA_DIR = PROJECT_ROOT / "Data" / "raw"
CLEANED_DATA_DIR = PROJECT_ROOT / "Data" / "cleaned"

CLEANED_DATA_DIR.mkdir(parents=True, exist_ok=True)

ORDERS_FILE = RAW_DATA_DIR / "olist_orders_dataset.csv"

df = pd.read_csv(ORDERS_FILE)

print("Dataset loaded successfully!")
```

```python
# Initial Inspection
print("Shape:", df.shape)

display(df.head())

print("\nData Types:")
display(df.dtypes)

print("\nColumns:")
print(df.columns.tolist())
```

```python
# missing values
missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

display(missing_report)
```

```python
# Duplicate Analysis
print("Full Duplicate Rows:", df.duplicated().sum())

print("Duplicate Order IDs:", df["order_id"].duplicated().sum())

print("Unique Order IDs:", df["order_id"].nunique())
```

```python
# Date Columns Analysis
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

display(df[date_columns].describe())
for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

display(df[date_columns].describe())
```

```python
# Order Status Analysis
print("Order Status Distribution:")

display(
    df["order_status"]
    .value_counts(dropna=False)
    .to_frame("Count")
)
```

```python
# Date Logic Validation
date_checks = {
    "Approval Before Purchase": (
        df["order_approved_at"] < df["order_purchase_timestamp"]
    ),

    "Carrier Before Purchase": (
        df["order_delivered_carrier_date"] < df["order_purchase_timestamp"]
    ),

    "Customer Delivery Before Purchase": (
        df["order_delivered_customer_date"] < df["order_purchase_timestamp"]
    ),

    "Customer Delivery Before Carrier": (
        df["order_delivered_customer_date"] < df["order_delivered_carrier_date"]
    ),

    "Delivery Before Approval": (
        df["order_delivered_customer_date"] < df["order_approved_at"]
    ),

    "Delivery After Estimated Date": (
        df["order_delivered_customer_date"] > df["order_estimated_delivery_date"]
    )
}

for check, condition in date_checks.items():
    print(f"{check}: {condition.sum()}")
```

```python
# Missing Values by Order Status
missing_by_status = (
    df.groupby("order_status")[date_columns]
    .apply(lambda group: group.isnull().sum())
)

display(missing_by_status)
```

```python
# Investigate Missing Dates
for col in date_columns:
    print(f"\n{col}")

    display(
        df[df[col].isnull()]["order_status"]
        .value_counts(dropna=False)
        .to_frame("Count")
    )
```

```python
# Create Cleaning Copy
cleaned_df = df.copy()

print("Original Shape:", df.shape)
print("Cleaning Copy Shape:", cleaned_df.shape)
```

```python
# Remove Exact Duplicates
before = cleaned_df.shape[0]

cleaned_df = cleaned_df.drop_duplicates()

after = cleaned_df.shape[0]

print("Removed Duplicate Rows:", before - after)
print("Remaining Rows:", after)
```

```python
# Validate Order IDs
print("Missing Order IDs:", cleaned_df["order_id"].isnull().sum())

print("Duplicate Order IDs:", cleaned_df["order_id"].duplicated().sum())
```

```python
# Final Data Types
print(cleaned_df.dtypes)
```

```python
# Final Quality Report

final_report = pd.DataFrame({
    "Column": cleaned_df.columns,
    "Data Type": cleaned_df.dtypes.astype(str).values,
    "Missing Values": cleaned_df.isnull().sum().values,
    "Missing Percentage": (
        cleaned_df.isnull().mean().values * 100
    )
})

display(final_report)

print("Final Shape:", cleaned_df.shape)
print("Duplicate Rows:", cleaned_df.duplicated().sum())
```

```python
# Save Cleaned Dataset
OUTPUT_FILE = CLEANED_DATA_DIR / "olist_orders_cleaned.csv"

cleaned_df.to_csv(OUTPUT_FILE, index=False)

print("Cleaned dataset saved successfully!")
print(OUTPUT_FILE)
```

```python
print("MISSING DATES BY ORDER STATUS")

for col in date_columns:
    print(f"\n{col}")
    display(
        df[df[col].isnull()]["order_status"]
        .value_counts(dropna=False)
        .to_frame("Count")
    )
```

```python
unusual_missing = df[
    (
        (df["order_status"] == "delivered") &
        (
            df["order_approved_at"].isnull() |
            df["order_delivered_customer_date"].isnull()
        )
    )
]

print("Unusual Missing Records:", len(unusual_missing))

display(unusual_missing)
```

```python
invalid_dates = df[
    (
        df["order_delivered_carrier_date"].notna() &
        (
            df["order_delivered_carrier_date"]
            < df["order_purchase_timestamp"]
        )
    ) |
    (
        df["order_delivered_customer_date"].notna() &
        df["order_approved_at"].notna() &
        (
            df["order_delivered_customer_date"]
            < df["order_approved_at"]
        )
    ) |
    (
        df["order_delivered_customer_date"].notna() &
        df["order_delivered_carrier_date"].notna() &
        (
            df["order_delivered_customer_date"]
            < df["order_delivered_carrier_date"]
        )
    )
]

print("Invalid Date Records:", len(invalid_dates))

display(invalid_dates)
```

```python
# Date Logic Validation 
date_issues = pd.DataFrame(index=df.index)

date_issues["Approval Before Purchase"] = (
    df["order_approved_at"] < df["order_purchase_timestamp"]
)

date_issues["Carrier Before Purchase"] = (
    df["order_delivered_carrier_date"] < df["order_purchase_timestamp"]
)

date_issues["Carrier Before Approval"] = (
    df["order_delivered_carrier_date"] < df["order_approved_at"]
)

date_issues["Delivery Before Approval"] = (
    df["order_delivered_customer_date"] < df["order_approved_at"]
)

date_issues["Delivery Before Carrier"] = (
    df["order_delivered_customer_date"] < df["order_delivered_carrier_date"]
)

print("Violations per Rule:")
display(date_issues.sum().to_frame("Count"))

print("Total Records with at least one violation:",
      date_issues.any(axis=1).sum())

print("Violations by Order Status:")

display(
    df.loc[date_issues.any(axis=1), "order_status"]
    .value_counts()
    .to_frame("Count")
)
```

```python
# Display Sample Records for Each Issue
for issue in date_issues.columns:
    print(f"\n{issue}")

    indexes = date_issues.index[date_issues[issue]]

    display(
        df.loc[indexes, [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date",
            "order_delivered_customer_date"
        ]].head(5)
    )
```

```python
issue = "Carrier Before Approval"

sample = df.loc[
    date_issues[issue],
    [
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date"
    ]
].head(10)

display(sample.to_string(index=True))
```

```python
# Calculate Time Difference for Carrier Before Approval
carrier_before_approval = df[
    date_issues["Carrier Before Approval"]
].copy()

carrier_before_approval["time_difference_hours"] = (
    carrier_before_approval["order_approved_at"]
    - carrier_before_approval["order_delivered_carrier_date"]
).dt.total_seconds() / 3600

display(
    carrier_before_approval["time_difference_hours"].describe()
)
```

```python
# Date Difference Distribution

bins = [-1, 1, 6, 24, 72, 168, float("inf")]

labels = [
    "Less than 1 hour",
    "1–6 hours",
    "6–24 hours",
    "1–3 days",
    "3–7 days",
    "More than 7 days"
]

carrier_before_approval["difference_group"] = pd.cut(
    carrier_before_approval["time_difference_hours"],
    bins=bins,
    labels=labels
)

display(
    carrier_before_approval["difference_group"]
    .value_counts()
    .sort_index()
    .to_frame("Records")
)
```

```python
# Create Working Copy

cleaned_df = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", cleaned_df.shape)
```

```python
# Clean Text Columns

text_columns = cleaned_df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].str.strip()

print("Text columns cleaned.")
```

```python
# Remove Exact Duplicate Rows

duplicates_before = cleaned_df.duplicated().sum()

cleaned_df = cleaned_df.drop_duplicates()

duplicates_after = cleaned_df.duplicated().sum()

print("Duplicates before:", duplicates_before)
print("Duplicates after:", duplicates_after)
print("Rows removed:", duplicates_before - duplicates_after)
```

```python
# Validate Order IDs

print("Missing Order IDs:", cleaned_df["order_id"].isna().sum())

print("Duplicate Order IDs:", cleaned_df["order_id"].duplicated().sum())

print("Unique Order IDs:", cleaned_df["order_id"].nunique())
```

```python
# Validate Order Status

valid_statuses = [
    "delivered",
    "shipped",
    "canceled",
    "unavailable",
    "invoiced",
    "processing",
    "created",
    "approved"
]

invalid_statuses = cleaned_df[
    ~cleaned_df["order_status"].isin(valid_statuses)
]

print("Invalid Status Records:", len(invalid_statuses))

display(invalid_statuses.head())
```

```python
# Date Logic Validation

date_issues = pd.DataFrame(index=cleaned_df.index)

date_issues["Carrier Before Purchase"] = (
    cleaned_df["order_delivered_carrier_date"].notna() &
    cleaned_df["order_purchase_timestamp"].notna() &
    (
        cleaned_df["order_delivered_carrier_date"] <
        cleaned_df["order_purchase_timestamp"]
    )
)

date_issues["Carrier Before Approval"] = (
    cleaned_df["order_delivered_carrier_date"].notna() &
    cleaned_df["order_approved_at"].notna() &
    (
        cleaned_df["order_delivered_carrier_date"] <
        cleaned_df["order_approved_at"]
    )
)

date_issues["Delivery Before Approval"] = (
    cleaned_df["order_delivered_customer_date"].notna() &
    cleaned_df["order_approved_at"].notna() &
    (
        cleaned_df["order_delivered_customer_date"] <
        cleaned_df["order_approved_at"]
    )
)

date_issues["Delivery Before Carrier"] = (
    cleaned_df["order_delivered_customer_date"].notna() &
    cleaned_df["order_delivered_carrier_date"].notna() &
    (
        cleaned_df["order_delivered_customer_date"] <
        cleaned_df["order_delivered_carrier_date"]
    )
)

display(date_issues.sum().to_frame("Violations"))

print(
    "Records with at least one date violation:",
    date_issues.any(axis=1).sum()
)
```

```python
# Build Data Quality Report

quality_checks = {
    "Total Records": len(cleaned_df),
    "Total Columns": len(cleaned_df.columns),
    "Exact Duplicate Rows": cleaned_df.duplicated().sum(),
    "Missing Order IDs": cleaned_df["order_id"].isna().sum(),
    "Duplicate Order IDs": cleaned_df["order_id"].duplicated().sum(),
    "Invalid Order Status": len(invalid_statuses),
    "Records with Date Violations": date_issues.any(axis=1).sum(),
    "Late Deliveries": (
        (
            cleaned_df["order_delivered_customer_date"] >
            cleaned_df["order_estimated_delivery_date"]
        ).sum()
    )
}

quality_report = pd.DataFrame(
    list(quality_checks.items()),
    columns=["Quality Check", "Result"]
)

display(quality_report)
```

```python
# Export Date Issue Records

date_issue_records = cleaned_df.loc[
    date_issues.any(axis=1)
].copy()

date_issue_records["has_date_issue"] = True

for col in date_issues.columns:
    date_issue_records[col] = date_issues.loc[
        date_issue_records.index, col
    ]

print("Date issue records:", len(date_issue_records))

display(date_issue_records.head())
```

```python
# Final Dataset Validation

print("Original Records:", len(df))
print("Cleaned Records:", len(cleaned_df))

print("Original Columns:", len(df.columns))
print("Cleaned Columns:", len(cleaned_df.columns))

print("Remaining Exact Duplicates:", cleaned_df.duplicated().sum())

print(
    "Unique Order IDs:",
    cleaned_df["order_id"].nunique()
)

print(
    "Missing Order IDs:",
    cleaned_df["order_id"].isna().sum()
)

print("\nFinal Data Types:")
display(cleaned_df.dtypes)
```

```python
# Save Cleaned Dataset and Quality Reports

CLEANED_DATA_DIR.mkdir(parents=True, exist_ok=True)

cleaned_file = CLEANED_DATA_DIR / "olist_orders_cleaned.csv"

quality_file = CLEANED_DATA_DIR / "orders_quality_report.csv"

date_issues_file = CLEANED_DATA_DIR / "orders_date_issues.csv"

cleaned_df.to_csv(cleaned_file, index=False)

quality_report.to_csv(quality_file, index=False)

date_issue_records.to_csv(date_issues_file, index=False)

print("Files saved successfully:")

print(cleaned_file)
print(quality_file)
print(date_issues_file)
```

```python
# Verify Saved Dataset

saved_df = pd.read_csv(
    cleaned_file,
    parse_dates=date_columns
)

print("Saved Dataset Shape:", saved_df.shape)

print("Expected Dataset Shape:", cleaned_df.shape)

assert saved_df.shape == cleaned_df.shape

assert saved_df["order_id"].nunique() == cleaned_df["order_id"].nunique()

print("Saved dataset verification passed.")
```
