# Customers Exploratory Data Analysis (EDA)

## Objective
Analyze the Olist customers dataset to evaluate dataset structure, identifier uniqueness, missing values, duplicates, and geographic distribution across cities and states.

## Dataset
`olist_customers_dataset.csv`

```python
import pandas as pd
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

df_customers = pd.read_csv(RAW_DATA_DIR / "olist_customers_dataset.csv")
df_customers.head()
```

## Data Structure & Overview
Review total records, feature names, and data types.

```python
df_customers.info()
```

## Data Integrity & Missing Values
Check for null values, duplicates, and primary key uniqueness.

```python
print("Null Values:")
print(df_customers.isnull().sum())
print("\nDuplicate Rows:", df_customers.duplicated().sum())
print("\nUnique customer_id:", df_customers['customer_id'].nunique())
print("Unique customer_unique_id:", df_customers['customer_unique_id'].nunique())
```

## Geographic Distribution Analysis
Identify top customer concentration by states and cities.

```python
print("Top 10 States:")
print(df_customers['customer_state'].value_counts().head(10))

print("\nTop 10 Cities:")
print(df_customers['customer_city'].value_counts().head(10))
```

## Zip Code Prefix Analysis
Analyze customer distribution across zip code prefixes.

```python
print("Unique Zip Code Prefixes:", df_customers['customer_zip_code_prefix'].nunique())
print("\nTop 10 Zip Code Prefixes:")
print(df_customers['customer_zip_code_prefix'].value_counts().head(10))
```

## Key Findings Summary
- Total Records & Integrity: 99,441 rows with 0 missing values and 0 duplicate rows across all columns.
- Identifier Uniqueness: `customer_id` is unique per order (99,441), while `customer_unique_id` represents unique individual buyers (96,096), revealing repeat purchasing behavior (~3.36% repeat rate).
- Geographic Distribution: High customer concentration in São Paulo state (`SP` with 41,746 customers) and São Paulo city (15,540), followed by Rio de Janeiro (`RJ`) and Minas Gerais (`MG`).
- Postal Distribution: 14,994 unique zip code prefixes identified, with top customer density concentrated in Rio de Janeiro coastal prefixes (e.g., `22790`, `24220`).


- Data Cleaning Note: `customer_zip_code_prefix` is currently stored as `int64`. Leading zeros (e.g., `01001`) should be padded (`zfill(5)`) during data cleaning to preserve valid 5-digit Brazilian postal codes.
